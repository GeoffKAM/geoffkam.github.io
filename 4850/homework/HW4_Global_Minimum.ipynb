### Question 4: Simulated annealing (3 marks)

All of the methods above are local minimization methods, and they perform poorly when the function is highly irregular. In such cases, a safer choice is a global minimization method.

One of the most common is simulated annealing (SA), an optimization algorithm inspired by the annealing process in metallurgy. It is a probabilistic technique for approximating the global optimum of a given function. Unlike deterministic methods such as gradient descent or Newton's method, SA occasionally accepts "bad" moves, which lets it escape local minima and explore the search space more effectively.

SA minimizes an objective function $f(x)$ by searching for an optimal $x$ with a temperature-based probabilistic rule.

#### Algorithm

1. **Initialize.**
   - Choose an initial solution $x$.
   - Set an initial temperature $T_0$ (e.g. 10).
   - Choose a cooling rate $\alpha$ (e.g. 0.99).
   - Choose a minimum temperature $T_{\min}$ for the stopping rule (e.g. $10^{-3}$).

2. **Iterate until $T < T_{\min}$.**
   1. Generate a new candidate solution $x_{\text{new}}$ by randomly perturbing $x$, for example by adding a draw from $U[-1, 1]$.
   2. Compute the change in the objective function:
      $$\Delta E = f(x_{\text{new}}) - f(x).$$
   3. Apply the acceptance rule:
      - If $\Delta E < 0$ ($x_{\text{new}}$ improves the objective), accept $x_{\text{new}}$.
      - If $\Delta E \geq 0$ ($x_{\text{new}}$ worsens the objective), accept it with probability
        $$P = e^{-\Delta E / T}.$$
      - This rule makes worse solutions more likely to be accepted at higher temperatures, which promotes exploration.
   4. Update the best solution found so far.
   5. Lower the temperature using the cooling schedule
      $$T \leftarrow \alpha T.$$
   6. Repeat until the temperature falls below $T_{\min}$.

3. **Return the best solution** found.
