### Question 1: Simulating $\pi$ (3 marks)

A classic exercise is to approximate $\pi$ by Monte Carlo simulation. Imagine a circle of radius $r = 0.5$ drawn inside the unit square, with its centre at $(x, y) = (0.5, 0.5)$. If you draw two uniform random numbers on $[0, 1]$ as the $x$ and $y$ coordinates of a point, the probability that the point falls inside the circle is the ratio of the two areas, $\pi r^2 / (1 \times 1)$. A point lies inside the circle exactly when its Euclidean distance from the centre is less than $0.5$, and we can use this fact to simulate $\pi$. More generally, simulation is a common way to compute a quantity of interest when we know a property that characterises it.

3. Use a `for` loop to count the number of pairs that fall inside the circle.

### Question 2: Faster Fibonacci numbers with dynamic programming (4 marks)

The recursive structure of the Fibonacci sequence, $F_n = F_{n-1} + F_{n-2}$, makes it a perfect example of code that works backwards recursively from $F_n$. Such code is also easy to write.

However, the naive recursive implementation (as in the lecture) is slow: its complexity is $\mathcal{O}(2^n)$. We can make it faster by tabulating the results forwards:

1. Start from the first two Fibonacci numbers, $(0, 1)$.
2. Use the current pair as your information (the current state) and compute the next number, $0 + 1 = 1$.
3. Update your information, $(0, 1) \rightarrow (1, 1)$, compute the next number, $1 + 1 = 2$, update again, $(1, 1) \rightarrow (1, 2)$, and keep iterating forwards until you reach the $n$-th iteration.

This approach is called tabulation, or bottom-up dynamic programming. Many economic problems are solved with dynamic programming. The complexity becomes $\mathcal{O}(n)$.

Implement both methods: the naive recursive function and the bottom-up function. For $n = 2, 5, 8, 11, 14$, record the computation time of each with `time = %timeit -o your_function(n)` (the mean time in seconds is `time.average`), and compute the ratio of the recursive time to the bottom-up time. Plot the ratios against $n$ using `matplotlib.pyplot.plot`. How does the ratio increase with $n$? Explain.
