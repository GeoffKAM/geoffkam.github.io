### Question 1: Solving an ordinary differential equation (5 marks, difficult)

A beautiful idea from Newton is that it is sometimes more elegant to describe a relationship by its rate of change than by the whole function. A well-known example in economics is the analysis of the economic impact of COVID-19 with the [SIR model](https://en.wikipedia.org/wiki/Compartmental_models_in_epidemiology).

Solving a full SIR model, and the theory of differential equations, are beyond the scope of this course. However, it is fairly easy to solve a first-order ordinary differential equation (ODE) numerically. Our objective is to find $y = g(x)$.

A first-order ODE has the form

$$\frac{dy}{dx} = f(y, x).$$

This only describes the rate of change, so you also need a level to start from. We call that level the initial (or boundary) condition.

Recall that we can approximate a derivative by a finite difference over a grid of $x$ values of size `N` (nothing is continuous in a computer), $x_0, x_1, \dots, x_{N-1}$. Then

$$\frac{dy}{dx} = f(y, x) \approx \frac{y_{i+1} - y_i}{x_{i+1} - x_i}.$$

The initial condition gives you $y_0$, and you know $x_0$ to $x_{N-1}$. The remaining question is what to substitute for $f(y, x)$ in order to find $y_{i+1}$ for all $i \in \{0, 1, \dots, N-2\}$. There are several methods.

#### Explicit method

Evaluate $f$ at the current grid point:

$$f(y_i, x_i) = \frac{y_{i+1} - y_i}{x_{i+1} - x_i}.$$

You can rearrange this and solve for all $y_i$ iteratively with a simple loop.

#### Implicit method

Evaluate $f$ at the next grid point:

$$f(y_{i+1}, x_{i+1}) = \frac{y_{i+1} - y_i}{x_{i+1} - x_i}.$$

The unknown $y_{i+1}$ appears on both sides, so you have to find $y_{i+1}$ as a root, using a root-finder such as `scipy.optimize.root`.

#### Implicit method with one Newton step

Finding the exact root at every grid point can be costly. An alternative is to apply Newton's method **once** at each grid point (instead of letting it converge) and use the implied root as $y_{i+1}$. Since this method is more difficult, here is a hint. Define

$$F(y) = f(y, x_{i+1}) - \frac{y - y_i}{x_{i+1} - x_i}.$$

That is the only hint.
