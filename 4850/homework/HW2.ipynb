### Question 1: Solving first-order conditions with Newton's method (4 marks)

In the lecture we used SciPy to solve a constrained maximization problem. It is also easy to solve it with your own implementation of Newton's method, and this is the most common use of Newton's method in economics. The problem is

$$\max_{x_1, x_2} \; x_1^{1-\alpha} x_2^{\alpha}$$

subject to the budget constraint

$$p_1 x_1 + p_2 x_2 \leq m,$$

where $p_1 = 2$, $p_2 = 1$, $m = 10$ and $\alpha = 0.5$.

The first-order conditions (FOCs) form a system of equations $\mathbf{F}(\mathbf{x}^*) = \mathbf{0}$, where $\mathbf{x}^* = (x_1, x_2, \lambda)$ and $\lambda$ is the Lagrange multiplier, so we can use Newton's method to find the solution. Use `tol=1e-10`.

Because the problem is well behaved, the method should converge for any initial guess that does not violate the budget constraint. You may use the initial guess $(1, 1, 1)$.

Pass the parameters to your functions using an object created with `namedtuple`.
