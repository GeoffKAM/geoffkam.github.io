## 1. Introduction and Motivation

- The simulated method of moments (SMM) is an estimation technique used mainly when a model's moments are too complex, or even impossible, to derive analytically, as in dynamic stochastic models or nonlinear systems.

- It extends the traditional method of moments (MM) by using simulation to approximate the model's theoretical moments, which are then compared with the moments observed in the data.

## 2. Theoretical Background

### 2.1. The Method of Moments (MM)

- The classical MM estimator rests on the idea that the parameters of a model can be identified by matching the model's theoretical moments (e.g., means, variances) with the corresponding sample moments.

- Suppose the model implies a vector of moments $m(\theta)$, where $\theta$ is the parameter vector, and the data provide sample moments $\hat{m}$.

- In the simplest case, we set

$$
m(\theta) = \hat{m}
$$

and solve for $\theta$.

### 2.2. Extending MM via Simulation

- In many complex models, $m(\theta)$ cannot be obtained in closed form.

- Instead, we simulate the model to generate a synthetic dataset and compute the corresponding moments $m_{\text{sim}}(\theta)$.

- The goal is then to find the parameter vector $\theta$ that minimizes the distance between the simulated moments and the observed data moments.

## 3. Mathematical Formulation of SMM

### 3.1. Objective Function

- SMM estimation minimizes an objective function that measures the distance between the observed moments and the simulated moments.

- A common choice is the weighted least squares criterion

$$
Q(\theta) = \left[\hat{m} - m_{\text{sim}}(\theta)\right]^\prime W \left[\hat{m} - m_{\text{sim}}(\theta)\right],
$$

where:
- $\hat{m}$ is the vector of moments computed from the data;
- $m_{\text{sim}}(\theta)$ is the vector of moments computed from data simulated at parameter value $\theta$;
- $W$ is a positive definite weighting matrix that can be chosen (or estimated) to improve efficiency.

### 3.2. The SMM Estimator

The SMM estimator $\hat{\theta}$ is the parameter value that minimizes the objective function:

$$
\hat{\theta} = \arg \min_{\theta} Q(\theta).
$$

## 4. Estimation Procedure

In practice, SMM generally proceeds in the following steps:

1. **Selection of moments**
   - Choose a set of moments (e.g., means, variances, autocorrelations) that capture the essential features of the data.
   - This choice is critical: the moments should be informative about the underlying parameters.

2. **Simulation**
   - For a candidate parameter vector $\theta$, simulate the model to generate synthetic data.
   - Compute the simulated moments $m_{\text{sim}}(\theta)$ from the synthetic data.
   - A large number of simulation draws is often used to reduce simulation error.

3. **Evaluation of the objective function**
   - Compute the difference $\hat{m} - m_{\text{sim}}(\theta)$ and evaluate the objective function $Q(\theta)$.

4. **Optimization**
   - Use a numerical optimization method (e.g., gradient descent, Nelder–Mead or another derivative-free method) to search over the parameter space.
   - Find the parameter value $\hat{\theta}$ that minimizes $Q(\theta)$.

5. **Inference**
   - Under regularity conditions, the SMM estimator is consistent and asymptotically normal.
   - Inference about $\theta$ may require bootstrapping or other methods that account for simulation error.

## 5. Practical Considerations

### 5.1. Choice of Moments and Weighting Matrix

- **Moment selection:** choose moments that are sensitive to changes in the model parameters. In practice, moments that capture key features such as persistence, volatility or tail behavior are often used.

- **Identification:** you need at least as many moments as parameters, and each parameter must shift the moments in a way the others cannot replicate. For example, means alone cannot pin down a variance parameter; include a variance (or another dispersion moment) for it.

- **Weighting matrix:** an optimal weighting matrix $W$ (often the inverse of the asymptotic covariance matrix of the moments) can deliver efficiency gains.

    - However, a suboptimal choice is sometimes made for computational simplicity. We will use one in the homework.

### 5.2. Simulation Error and Computational Burden

- **Simulation error:** because the moments are computed by simulation, they contain simulation error. This error can be reduced by increasing the number of simulation draws.

- **Common random numbers:** draw the random shocks once (with a fixed seed) and reuse the same draws every time you evaluate $Q(\theta)$. If fresh shocks are drawn at each evaluation, $Q(\theta)$ jumps around even when $\theta$ barely changes, and the optimizer cannot settle on a minimum.

- **Computational demand:** SMM can be computationally intensive, particularly for high-dimensional models or when a very fine grid over the parameter space is needed. Efficient programming and parallel computing can ease some of these challenges.

### 5.3. Robustness and Model Validation

- **Robustness checks:** it is good practice to check how sensitive the estimates are to the choice of moments, the number of simulation draws and the weighting matrix.

- **Validation:** validate the fitted model with out-of-sample tests or other diagnostic checks to make sure the simulated moments adequately capture the features of the observed data.

- **Uniqueness:** because SMM relies on simulating the model's outcomes, it can only be used when those outcomes are uniquely determined for each parameter value. In other words, when you estimate a model by SMM, you need to show that the model predicts a unique equilibrium.
