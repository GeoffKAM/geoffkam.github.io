2. Python is free.
    - Almost everything is open source.
    - The existing infrastructure is excellent (e.g. VS Code).
    - It is likely that an existing library already solves your problem.
    - Learning materials are free as well.

3. Python is fast.
    - Despite being a high-level language, Python can be fast because:
        - Many core libraries are written in C.
        - It is easy to parallelize your code, both to write and to run.

4. Python is easy to learn and write.
    - Python is a high-level language, so humans can read it easily.

### Task 1: Simulate an AR(1) process

- Let's simulate an AR(1) process and calculate its moments (the mean and the variance over time). You may not understand every line yet; don't worry, we will come back to these ideas again and again.

- Simulate the following process for 140 periods:
$$y_{t+1}=\phi y_{t}+\epsilon_t,$$
where
    - $\epsilon_t \sim N(\mu,\sigma^2)$, independently over time,
    - $\mu = 0$,
    - $\phi = 0.9$,
    - $\sigma = 2$,
    - $y_0 = 0$.

- Plot the time path.

- Check whether the mean and variance of the simulated series are close to their theoretical values:
$$\operatorname{E}(y)=\frac{\mu}{1-\phi}, \qquad \operatorname{Var}(y)=\frac{\sigma^2}{1-\phi^2}.$$

```python
# numpy: numerical computation
# matplotlib: plotting
# pandas: displaying the results as a table
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Parameters
T = 140      # number of periods
phi = 0.9    # persistence
sigma = 2    # standard deviation of the shock
mu = 0       # mean of the shock
y_0 = 0      # initial value
```

```python
def generate_ar1(phi, sigma, mu, y_0, T):
    """Simulate T periods of the AR(1) process y_t = phi * y_{t-1} + epsilon_t."""
    # Fix the seed of numpy's random number generator so results are reproducible
    np.random.seed(850)

    # Space to store the path
    y = np.zeros(T)
    y[0] = y_0

    # Draw the whole sequence of shocks at once
    epsilon = np.random.normal(mu, sigma, T)

    # A for loop handles the repetitive calculation
    for t in range(1, T):
        y[t] = phi * y[t - 1] + epsilon[t]

    return y
```

```python
y = generate_ar1(phi, sigma, mu, y_0, T)

# Simulated moments
mean_simulated = np.mean(y)
variance_simulated = np.var(y)

# Theoretical moments
mean_theoretical = mu / (1 - phi)
variance_theoretical = sigma**2 / (1 - phi**2)

# Display the results as a table
moments = {
    "Moment": ["Mean", "Variance"],
    "Simulated": [mean_simulated, variance_simulated],
    "Theoretical": [mean_theoretical, variance_theoretical],
}
moments_df = pd.DataFrame(moments)
print(moments_df)

# Plot the time path
plt.figure(figsize=(10, 6))
plt.plot(np.arange(y.size), y, label='AR(1) Process', linewidth=2)
plt.axhline(mean_simulated, color='r', linestyle='--', label='Simulated Mean')
plt.axhline(mean_theoretical, color='g', linestyle='--', label='Theoretical Mean')
plt.title('Simulated AR(1) Process')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid()
plt.show()
```

The simulated moments are not very close to the theoretical ones. With only 140 periods of a persistent process ($\phi = 0.9$), the series contains few effectively independent observations, so its sample mean and variance are noisy; the path also starts at $y_0 = 0$ rather than at a draw from the long-run distribution. Increase `T` (say, to 100,000) and the simulated moments approach the theoretical values.
